```python
# This is critical — we need the EXACT column names
# Column names often have hidden spaces, different cases, or special characters

print("Column names in your dataset:")
for i, col in enumerate(df.columns):
    print(f"  {i+1}. '{col}'")
```

```python
print("Years covered:", sorted(df['Year'].unique()))
print("Rows per year:")
print(df['Year'].value_counts().sort_index())
```

```python
# .dtypes tells us what data type pandas assigned to each column
# This matters because:
# - 'Opening Rank' should be numeric. If it's 'object', something is wrong
# - 'Year' should be integer. If it's float, there may be missing values
# - Text columns should be 'object' type in pandas

print("Data types of each column:")
print(df.dtypes)
```

```python
# Let's formally confirm what pandas sees for every column
# object = text/string in pandas terminology

print("Column data types:")
print(df.dtypes)
print("\nNote: 'object' means pandas is treating that column as text")
```

```python
# This is one of the most important profiling steps
# .isnull().sum() counts how many blank/NaN values exist in each column
# We also calculate the percentage so we know the severity

missing_count = df.isnull().sum()
missing_percent = (df.isnull().sum() / len(df)) * 100

missing_profile = pd.DataFrame({
    'Missing Count': missing_count,
    'Missing %': missing_percent.round(2)
})

print("Missing Value Profile:")
print(missing_profile)
```

```python
# For categorical columns, we want to know how many unique values exist
# This tells us the cardinality (variety) of each column
# Example: 'Gender' should have only 2-3 unique values
# 'Institute' might have 100+ unique values

print("Unique values per column:")
for col in df.columns:
    print(f"  {col}: {df[col].nunique()} unique values")
```

```python
# Now we look at the ACTUAL unique values for text columns
# This reveals inconsistencies like 'OPEN' vs 'Open' vs 'open'
# or 'Female-Only' vs 'Female Only' vs 'F'

categorical_cols = ['Quota', 'Seat Type', 'Gender']

for col in categorical_cols:
    print(f"\n--- {col} ---")
    print(df[col].value_counts())
```

```python
# Institutes are the most complex categorical column
# We want to understand how many colleges, what types exist
# Since there is no 'Institute Type' column, we will extract it ourselves later

print(f"Total unique institutes: {df['Institute'].nunique()}")
print(f"\nFirst 20 institute names:")
for inst in sorted(df['Institute'].unique())[:20]:
    print(f"  {inst}")
```

```python
# .describe() gives us statistical summary of numeric columns
# For Opening Rank and Closing Rank, we want to see:
# - min value (should not be 0 or negative)
# - max value (rank of 1,000,000+ is unusual)
# - mean and median (if very different, data is skewed)

print("Statistical summary of numeric columns:")
df[['Opening Rank', 'Closing Rank', 'Round', 'Year']].describe()
```

```python
# Earlier we noticed 2021 had only 30,355 rows vs 50,000+ in other years
# Let's investigate WHY
# We check if certain rounds are missing for 2021

print("Round distribution by Year:")
round_year = df.groupby(['Year', 'Round']).size().unstack(fill_value=0)
print(round_year)
```

```python
# A fundamental rule: Opening Rank should ALWAYS be <= Closing Rank
# If Closing Rank < Opening Rank, that row is logically impossible
# We count how many such violations exist

invalid_ranks = df[df['Closing Rank'] < df['Opening Rank']]
print(f"Rows where Closing Rank < Opening Rank: {len(invalid_ranks)}")

if len(invalid_ranks) > 0:
    print("\nSample invalid rows:")
    print(invalid_ranks.head())
```
